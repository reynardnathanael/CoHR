```python
import sys
from data_loader import load_and_save
import os
import pandas as pd
# =========================
# PATH SETUP
# =========================
BASE_DIR = os.getcwd()
PROJECT_ROOT = os.path.abspath(os.path.join(BASE_DIR, ".."))

if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

from preprocessing.parser import parse_resume
```

```python
# add scripts folder so we can import
sys.path.append(os.path.abspath("../scripts"))
# run the loader
load_and_save("AzharAli05/Resume-Screening-Dataset", "azharali")
```

```python
# =========================
# LOAD DATA
# =========================
path = os.path.join(PROJECT_ROOT, "data", "raw", "azharali_train.csv")
df = pd.read_csv(path)

print(df.columns)
print(df.head())
```

```python
# =========================
# PARSE RESUMES
# =========================
parsed_rows = []

for i, row in df.iterrows():
    resume_text = row["Resume"]

    parsed = parse_resume(
        text=resume_text,
        resume_id=i,
    )

    # ADD EXTRA COLUMNS FROM ORIGINAL DATASET
    parsed["role"] = row["Role"] if "Role" in df.columns else None
    parsed["decision"] = row["Decision"] if "Decision" in df.columns else None
    parsed["reason_for_decision"] = row["Reason_for_decision"] if "Reason_for_decision" in df.columns else None

    parsed_rows.append(parsed)

parsed_df = pd.DataFrame(parsed_rows)
```

```python
# =========================
# CHECK RESULT
# =========================
print(parsed_df.head())

# =========================
# SAVE RESULT
# =========================
output_path = os.path.join(PROJECT_ROOT, "data", "raw", "azharali_train_parsed.csv")
parsed_df.to_csv(output_path, index=False)

print("Saved to:", output_path)
```
